```python
import math

def Trap(n, a, b, f) :
    h = (b - a) / n
    sum = (f(a) + f(b)) / 2
    for i in range(1, n) :
        x = a + i * h
        sum += f(x)
    return h * sum

upper_bound = math.pi * 2

def func(x) :
    a = math.sin(x) * math.sin(x)
    b = 5 + 4 * math.cos(x)

    return a / b


def conv_ratio(s) :
    a = Trap(2 * s, 0, upper_bound, func) - Trap(s, 0, upper_bound, func)
    b = Trap(4 * s, 0, upper_bound, func) - Trap(2 * s, 0, upper_bound, func)
    return (a / b)



print("Approximation with N = 4: ")
print(Trap(4, 0, upper_bound, func))
print("\n")

print("Approximation with N = 8: ")
print(Trap(8, 0, upper_bound, func))
print("\n")

print("Approximation with N = 16: ")
print(Trap(16, 0, upper_bound, func))
print("\n")

print("Approximation with N = 32: ")
print(Trap(32, 0, upper_bound, func))
print("\n")

print("Convergence rate:")
print(conv_ratio(32))
print("Pi/4 = ")
print(math.pi/4)
```

The function converges to pi/4 at such a quick rate because trapezoidal approximation is extremely effective for periodic functions.
